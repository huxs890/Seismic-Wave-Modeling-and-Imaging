*2D acoustic wave equation:*
$$\frac{\partial^2p}{\partial t^2}=v^2(x,z)\left(\frac{\partial^2p}{\partial x^2}+\frac{\partial^2p}{\partial z^2}\right)+s(x,z,t),$$

*Derivation of 2N-order finite difference in the spatial domain*

Taylor expansion

$$ 
p(x+\Delta x) - p(x) = \frac{\partial p}{\partial x}{\Delta x} + \frac{1}{2!}\frac{\partial^2 p}{\partial x^2}{\Delta x}^2 + \frac{1}{3!}\frac{\partial^3 p}{\partial x^3}{\Delta x}^3 + \dots + \frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}{\Delta x}^{2N} + O(\Delta x^{2N})  
$$
$$ 
p(x-\Delta x) - p(x) = -\frac{\partial p}{\partial x}{\Delta x} + \frac{1}{2!}\frac{\partial^2 p}{\partial x^2}{\Delta x}^2 - \frac{1}{3!}\frac{\partial^3 p}{\partial x^3}{\Delta x}^3 + \dots + \frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}{\Delta x}^{2N} + O(\Delta x^{2N})  
$$

$$\dots$$

$$
p(x+N\Delta x) - p(x) = \frac{\partial p}{\partial x}(N\Delta x) + \frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(N\Delta x)^2 + \frac{1}{3!}\frac{\partial^3 p}{\partial x^3}(N\Delta x)^3 + \dots + \frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}(N\Delta x)^{2N} + O(\Delta x^{2N})  
$$
$$ 
p(x+N\Delta x) - p(x) = -\frac{\partial p}{\partial x}(N\Delta x) + \frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(N\Delta x)^2 - \frac{1}{3!}\frac{\partial^3 p}{\partial x^3}(N\Delta x)^3 + \dots + \frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}(N\Delta x)^{2N} + O(\Delta x^{2N})  
$$

By adding the two expressions together in pairs, we obtain

$$
p(x+\Delta x) + p(x-\Delta x) - 2p(x) = 2\frac{1}{2!}\frac{\partial^2 p}{\partial x^2}{\Delta x}^2 + 2\frac{1}{4!}\frac{\partial^4 p}{\partial x^4}{\Delta x}^4 + \dots + 2\frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}{\Delta x}^{2N} + O(\Delta x^{2N})
$$
$$
p(x+2\Delta x) + p(x-2\Delta x) - 2p(x) = 2\frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(2\Delta x)^2 + 2\frac{1}{4!}\frac{\partial^4 p}{\partial x^4}(2\Delta x)^4 + \dots + 2\frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}(2\Delta x)^{2N} + O(\Delta x^{2N})
$$
$$\dots$$
$$
p(x+N\Delta x) + p(x-N\Delta x) - 2p(x) = 2\frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(N\Delta x)^2 + 2\frac{1}{4!}\frac{\partial^4 p}{\partial x^4}(N\Delta x)^4 + \dots + 2\frac{1}{(2N)!}\frac{\partial^{2N} p}{\partial x^{2N}}(N\Delta x)^{2N} + O(\Delta x^{2N})
$$

Multiply both sides of each resulting equation by the corresponding coefficient $C_1,C_2,...,C_M$, then add the left and right sides of all the equations togeter respectively, we can get
$$
\sum_{n=1}^{N}C_{n}\left[p(x+n\Delta x)+p(x-n\Delta x)-2p(x)\right]=
$$
$$
2C_1\frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(\Delta x)^2 + 
2C_2\frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(2\Delta x)^2 + 
\dots +
2C_N\frac{1}{2!}\frac{\partial^2 p}{\partial x^2}(N\Delta x)^2 + 
$$
$$
2C_1\frac{1}{4!}\frac{\partial^4 p}{\partial x^4}(\Delta x)^4 + 
2C_2\frac{1}{4!}\frac{\partial^4 p}{\partial x^4}(2\Delta x)^4 + 
\dots +
2C_N\frac{1}{4!}\frac{\partial^4 p}{\partial x^4}(N\Delta x)^4 + 
$$
$$\dots$$
$$
2C_1\frac{1}{2N!}\frac{\partial^{2N} p}{\partial x^{2N}}(\Delta x)^{2N} + 
2C_2\frac{1}{2N!}\frac{\partial^{2N} p}{\partial x^{2N}}(2\Delta x)^{2N} + 
\dots +
2C_N\frac{1}{2N!}\frac{\partial^{2N} p}{\partial x^{2N}}(N\Delta x)^{2N} + O(\Delta x^{2N})
$$

After simplifying the above equation, we obtain
$$\sum_{n=1}^{N}C_{n}\left[p(x+n\Delta x)+p(x-n\Delta x)-2p(x)\right]=$$
$$\frac{2}{2!}\left\{C_{1}+2^2 C_{2}+\cdots+N^2C_{N}\right\}\frac{\partial^2p}{\partial x^2}\Delta x^2+$$
$$\frac{2}{4!}\left\{C_{1}+2^4 C_{2}+\cdots+N^4C_{N}\right\}\frac{\partial^4p}{\partial x^4}\Delta x^4+$$
$$\dots$$
$$\frac{2}{2N!}\left\{C_{1}+2^{2N} C_{2}+\cdots+N^{2N}C_{N}\right\}\frac{\partial^{2N}p}{\partial x^{2N}}\Delta x^{2N}+O(\Delta x^{2N})$$

The spatial domain of the wave equation is a second-order partial differential format. Therefore, the coefficients of this equation can be set to 1, while the coefficients of other higher-order partial differential formats are set to 0. That is
$$C_{1}+2^2 C_{2}+\cdots+N^2C_{N}=1$$
$$C_{1}+2^4 C_{2}+\cdots+N^4C_{N}=0$$
$$\cdots$$
$$C_{1}+2^{2N} C_{2}+\cdots+N^{2N}C_{N}=0$$

Rewrite the above equations in matrix form
$$
\begin{bmatrix}
1^2 & 2^2 & 3^2 & \cdots & N^2 \\
1^4 & 2^4 & 3^4 & \cdots & N^4 \\
1^6 & 2^6 & 3^6 & \cdots & N^6 \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
1^{2N} & 2^{2N} & 3^{2N} & \cdots & N^{2N}
\end{bmatrix}
\begin{bmatrix}
C_1 \\
C_2 \\
C_3 \\
\vdots \\
C_N
\end{bmatrix}
=
\begin{bmatrix}
1 \\
0 \\
0 \\
\vdots \\
0
\end{bmatrix}
$$

Through the above derivation, we have obtained a 2N-order difference scheme of $\frac{\partial^2 p}{\partial x^2}$,
$$
\frac{\partial^2 p}{\partial x^2} = \frac{1}{\Delta x^2} \sum_{n=1}^{N}C_{n}\left[p(x+n\Delta x)+p(x-n\Delta x)-2p(x)\right]
$$

Other difference schemes can also be constructed in this way.

*The numerical solution formula for the high-order difference scheme of the acoustic wave equation*
$$
\frac{\partial^2p}{\partial t^2}=v^2(x,z)\left(\frac{\partial^2p}{\partial x^2}+\frac{\partial^2p}{\partial z^2}\right)+s(x,z,t)
$$

Where

$$
\frac{\partial^2 p}{\partial x^2} = \frac{1}{\Delta x^2} \sum_{n=1}^{N}C_{n}\left[p(x+n\Delta x)+p(x-n\Delta x)-2p(x)\right]
$$
$$
\frac{\partial^2 p}{\partial z^2} = \frac{1}{\Delta z^2} \sum_{n=1}^{N}C_{n}\left[p(z+n\Delta z)+p(z-n\Delta z)-2p(z)\right]
$$
$$
\frac{\partial^2 p}{\partial t^2} = \frac{1}{\Delta t^2} \left[p(t+\Delta t)+p(t-\Delta t)-2p(t)\right]
$$

Substituting the above difference scheme into the acoustic wave equation yields
$$
\frac{1}{\Delta t^2} \left[p(t+\Delta t)+p(t-\Delta t)-2p(t)\right] = 
\frac{v^2(x,z)}{\Delta x^2} \sum_{n=1}^{N}C_{n}\left[p(x+n\Delta x)+p(x-n\Delta x)-2p(x)\right] + \\
\frac{v^2(x,z)}{\Delta z^2} \sum_{n=1}^{N}C_{n}\left[p(z+n\Delta z)+p(z-n\Delta z)-2p(z)\right] + s(x,z,t)
$$

Explicit difference scheme

$$
\begin{aligned}
p(t+\Delta t) = &
2p(t)-p(t-\Delta t) \\
& +\frac{v^2(x,z) \Delta t^2}{\Delta x^2} \sum_{n=1}^{N}C_{n}\left[p(x+n\Delta x)+p(x-n\Delta x)-2p(x)\right] \\
& +\frac{v^2(x,z) \Delta t^2}{\Delta z^2} \sum_{n=1}^{N}C_{n}\left[p(z+n\Delta z)+p(z-n\Delta z)-2p(z)\right] \\
& +\Delta t^2 s(x,z,t)
\end{aligned}
$$
