*Staggered-grid first-order pressure–velocity formulation*

$$\begin{aligned}&\frac{\partial v_x}{\partial t}=-\frac{1}{\rho}\frac{\partial p}{\partial x},\\&\frac{\partial v_z}{\partial t}=-\frac{1}{\rho}\frac{\partial p}{\partial z},\\&\frac{\partial p}{\partial t}=-\rho v^2\left(\frac{\partial v_x}{\partial x}+\frac{\partial v_z}{\partial z}\right)+s(x,z,t)\end{aligned}$$

Taylor expansion
$$p\left(x+\frac{1}{2}\Delta x\right)=p(x)+\frac{\partial p}{\partial x}(\frac{\Delta x}{2}) + \frac{1}{2!}\frac{\partial^2p}{\partial x^2}\left(\frac{\Delta x}{2}\right)^2+\cdots+\frac{1}{(2N)!}\frac{\partial^{2N}p}{\partial x^{2N}}\left(\frac{\Delta x}{2}\right)^{2N} + O(\Delta x^{2N})$$
$$p\left(x-\frac{1}{2}\Delta x\right)=p(x)-\frac{\partial p}{\partial x}(\frac{\Delta x}{2}) + \frac{1}{2!}\frac{\partial^2p}{\partial x^2}\left(\frac{\Delta x}{2}\right)^2+\cdots+\frac{1}{(2N)!}\frac{\partial^{2N}p}{\partial x^{2N}}\left(\frac{\Delta x}{2}\right)^{2N} + O(\Delta x^{2N})$$
$$\cdots$$
$$p\left[x+(N-\frac{1}{2})\Delta x\right]=p(x)+\frac{\partial p}{\partial x}((N-\frac{1}{2})\Delta x) + \frac{1}{2!}\frac{\partial^2p}{\partial x^2}\left((N-\frac{1}{2})\Delta x\right)^2+\cdots+\frac{1}{(2N)!}\frac{\partial^{2N}p}{\partial x^{2N}}\left((N-\frac{1}{2})\Delta x\right)^{2N} + O(\Delta x^{2N})$$
$$p\left[x-(N-\frac{1}{2})\Delta x\right]=p(x)-\frac{\partial p}{\partial x}((N-\frac{1}{2})\Delta x) + \frac{1}{2!}\frac{\partial^2p}{\partial x^2}\left((N-\frac{1}{2})\Delta x\right)^2+\cdots+\frac{1}{(2N)!}\frac{\partial^{2N}p}{\partial x^{2N}}\left((N-\frac{1}{2})\Delta x\right)^{2N} + O(\Delta x^{2N})$$

Then
$$
p\left(x+\frac{1}{2}\Delta x\right)-p\left(x-\frac{1}{2}\Delta x\right) = 2\frac{\partial p}{\partial x}(\frac{\Delta x}{2}) + 2\frac{\partial^3 p}{\partial x^3}(\frac{\Delta x}{2})^3 + \cdots + 2\frac{\partial^{2N-1} p}{\partial x^{2N-1}}(\frac{\Delta x}{2})^{2N-1} + O(\Delta x^{2N-1})
$$
$$
p\left(x+\frac{3}{2}\Delta x\right)-p\left(x-\frac{3}{2}\Delta x\right) = 2\frac{\partial p}{\partial x}(\frac{3\Delta x}{2}) + 2\frac{\partial^3 p}{\partial x^3}(\frac{3\Delta x}{2})^3 + \cdots + 2\frac{\partial^{2N-1} p}{\partial x^{2N-1}}(\frac{3\Delta x}{2})^{2N-1} + O(\Delta x^{2N-1})
$$
$$
\cdots
$$
$$
p\left(x+(N-\frac{1}{2})\Delta x\right)-p\left(x-(N-\frac{1}{2})\Delta x\right) = 2\frac{\partial p}{\partial x}(N-\frac{1}{2})\Delta x + 2\frac{\partial^3 p}{\partial x^3}((N-\frac{1}{2})\Delta x)^3 + \cdots + 2\frac{\partial^{2N-1} p}{\partial x^{2N-1}}((N-\frac{1}{2})\Delta x)^{2N-1} + O(\Delta x^{2N-1})
$$

Then
$$
\sum_{n=1}^{N}C_{n}\left\{p\left[x+\left(n-\frac{1}{2}\right)\Delta x\right]-p\left[x-\left(n-\frac{1}{2}\right)\Delta x\right]\right\} = 
$$
$$
2C_1\frac{\partial p}{\partial x}(\frac{\Delta x}{2}) + 2C_2\frac{\partial p}{\partial x}(\frac{3\Delta x}{2}) + \cdots + 2C_N\frac{\partial p}{\partial x}(N-\frac{1}{2})\Delta x
$$
$$
+ 2C_1\frac{\partial^3 p}{\partial x^3}(\frac{\Delta x}{2})^3 + 2C_2\frac{\partial^3 p}{\partial x^3}(\frac{3\Delta x}{2})^3 + \cdots + 2C_N\frac{\partial^3 p}{\partial x^3}((N-\frac{1}{2})\Delta x)^3
$$
$$\cdots$$
$$
+ 2C_1\frac{\partial^{2N-1} p}{\partial x^{2N-1}}(\frac{\Delta x}{2})^{2N-1} + 2C_2\frac{\partial^{2N-1} p}{\partial x^{2N-1}}(\frac{3\Delta x}{2})^{2N-1}+ \cdots + 2C_N\frac{\partial^{2N-1} p}{\partial x^{2N-1}}((N-\frac{1}{2})\Delta x)^{2N-1}
$$

Then
$$
\sum_{n=1}^{N}C_{n}\left\{p\left[x+\left(n-\frac{1}{2}\right)\Delta x\right]-p\left[x-\left(n-\frac{1}{2}\right)\Delta x\right]\right\} = 
$$
$$
\left[2(\frac{1}{2})C_1 + 2(\frac{3}{2})C_2 + \cdots + 2(N-\frac{1}{2})C_N\right]\frac{\partial p}{\partial x}\Delta x
$$
$$
+\left[2(\frac{1}{2})^3C_1 + 2(\frac{3}{2})^3C_2 + \cdots + 2(N-\frac{1}{2})^3C_N\right]\frac{\partial^3 p}{\partial x^3}\Delta x^3
$$
$$\cdots$$
$$
+\left[2(\frac{1}{2})^{2N-1}C_1 + 2(\frac{3}{2})^{2N-1}C_2 + \cdots + 2(N-\frac{1}{2})^{2N-1}C_N\right]\frac{\partial^{2N-1} p}{\partial x^{2N-1}}\Delta x^{2N-1}
$$

Then
$$\left[2(\frac{1}{2})C_1 + 2(\frac{3}{2})C_2 + \cdots + 2(\frac{2N-1}{2})C_N\right]=1$$
$$\left[2(\frac{1}{2})^3C_1 + 2(\frac{3}{2})^3C_2 + \cdots + 2(\frac{2N-1}{2})^3C_N\right]=0$$
$$\cdots$$
$$\left[2(\frac{1}{2})^{2N-1}C_1 + 2(\frac{3}{2})^{2N-1}C_2 + \cdots + 2(\frac{2N-1}{2})^{2N-1}C_N\right]=0$$

Then
$$\left[1C_1 + 3C_2 + \cdots + (2N-1)C_N\right]=1$$
$$\left[1^3C_1 + 3^3C_2 + \cdots + (2N-1)^3C_N\right]=0$$
$$\cdots$$
$$\left[1^{2N-1}C_1 + 3^{2N-1}C_2 + \cdots + (2N-1)^{2N-1}C_N\right]=0$$

Then
$$\begin{bmatrix}1&3&5&\cdots&2N-1\\1^3&3^3&5^3&\cdots&(2N-1)^3\\1^5&3^5&5^5&\cdots&(2N-1)^5\\\vdots&\vdots&\vdots&\ddots&\vdots\\1^{2N-1}&3^{2N-1}&5^{2N-1}&\cdots&(2N-1)^{2N-1}\end{bmatrix}\begin{bmatrix}C_1\\C_2\\C_3\\\vdots\\C_N\end{bmatrix}=\begin{bmatrix}1\\0\\0\\\vdots\\0\end{bmatrix}$$
